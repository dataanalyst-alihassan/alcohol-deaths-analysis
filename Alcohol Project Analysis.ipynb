```python
import numpy as np

countries=np.array(["Pakistan","India","Russia",'Japan',"China",
                   "Thailand","South Korea","Vietnam"])
death=np.array([0.5,15.2,25.8,4.1,6.3,
                    14.3,9.2,11.8])
years=np.array([2018,2019,2020,2021,2022])

print("="*40)
print("ALCOHOL RELATED DEATHS ANALYSIS")
print("="*40)
print("Total Countries:",len(countries))
print("Countries:",countries)
print("Death Rate:",death)
#Applying function to analyze death rate

print("="*40)
print("APPLYING FUNCTIONS TO ANALYZE DEATH RATE")
print("="*40)

print("Total:",round(np.sum(death),2))
print("Average:",round(np.mean(death),2))
print("Median:", np.median(death))
print("Highest:", np.max(death))
print("Lowest:", np.min(death))
print("Std Dev:", round(np.std(death),2))
print("Variance:", round(np.var(death),2))

#Finding the index of the country with highest and lowest death rates 

max_idx = np.argmax(death)
min_idx = np.argmin(death)

print("="*40)
print("RATE ANALYSIS")
print("="*40)

# Use index to get COUNTRY NAME and DEATH RATE!
print("Highest Death Rate Country:", 
       countries[max_idx], "->", death[max_idx])

print("Lowest Death Rate Country:", 
       countries[min_idx], "->", death[min_idx])

#Risk Analysis
#1. High Risk countries (death > 12)
   #Print country names and death rates

#2. Low Risk countries (death < 8)
  # Print country names and death rates

#3. Medium Risk countries (between 8 and 12)
  # Print country names and death rates
print("="*40)
print("RISK ANALYSIS")
print("="*40)

high=np.where(death>12)
print("High Risk Countries:",countries[high])
print("Death Rates:",death[high])

low=np.where(death<8)
print("Low Risk Countries:",countries[low])
print("Death Rates:",death[low])

medium=np.where((death>8)&(death<12))
print('Medium Risk Countries',countries[medium])
print("Death Rates:",death[medium])  

#Now Labeling every country as High/Medium/Low Risk
print("="*40)
print("LABELING EVERY COUNTRY")
print("="*40)
label=np.where(death>12,"High Risk",
      np.where(death>=8,"Medium Risk","Low Risk"))
for i in range(len(countries)):
 print(countries[i],"->",death[i],'->',label[i])

print("="*40)
print("PERCENTILE ANALYSIS")
print("="*40)

print("25th Percentile:",np.percentile(death,25))
print("50th Percentile:",np.percentile(death,50))
print("75th Percentile:",np.percentile(death,75))
print("90th Percentile",np.percentile(death,90))

print("="*40)
print("COUNTRIES SORTED BY DEATH RATE")
print("="*40)

sorted_idx=np.argsort(death)

for i in sorted_idx:
    print(countries[i],"->",death[i])
```
